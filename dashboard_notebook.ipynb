```python
import pandas as pd
dash = pd.read_csv('dashboard_data.csv')
dash['admit_rate_residual'] = pd.to_numeric(dash['admit_rate_residual'], errors='coerce')
```

```python
# question: which ca public school type beats its expected uc admit rate most, 2022-2024
# admit_rate_residual = real rate minus expected rate (expected already controls for poverty, gpa, size)
d = dash[(dash['fall_term'].between(2022, 2024)) & (dash['school_type'].notna()) & (dash['campus'] == 'Universitywide')]
pub = d[d['school_type'].notna()]
```

```python
# average residual per school type, highest first
res = pub.groupby('school_type')['admit_rate_residual'].mean()
print(res.sort_values(ascending=False))
```

```python
# in percentage points, for the writeup
print((res * 100).round(1).sort_values(ascending=False))
```
